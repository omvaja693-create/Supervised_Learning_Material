```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
house_price_data = {
    'house_sqr_ft': [1200, 1500, 1800, 900, 2000],
    'price': [200000, 250000, 300000, 160000, 330000]
}

df = pd.DataFrame(house_price_data)
df
```

```python
X = df[['house_sqr_ft']]
y = df['price']
```

```python
# Create a model object
# Train that model
# Predict from that model
# Evaluate that model
```

```python
# Create a model object

lr = LinearRegression()
```

```python
# Train that model

lr.fit(X, y)
```

```python
# Predict from that model

y_pred = lr.predict([[1000]])[0]
print(y_pred)
```

```python
y_pred_ = lr.predict(X)
y_pred_
```

```python
# Evaluate that model

score = r2_score(y, y_pred_)

print(score)
```

```python
# Find the slope & intercept

m = lr.coef_[0]
c = lr.intercept_

print(f"Slope: {m}")
print(f"Intercept: {c}")
```
